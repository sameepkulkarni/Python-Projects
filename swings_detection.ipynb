```python
import pandas as pd
gold_csv = pd.read_csv("gold_250k_data.csv", index_col=0)
print(gold_csv.columns)
gold_csv.head(5)
```

```python

def detect_swing_highs(df, window=5):
    """
    Detect swing highs in price data
    window: number of bars to look before and after (default=5)
    """
    highs = []
    
    # Loop through the data
    for i in range(window, len(df) - window):
        # Get the current high
        current_high = df['h'].iloc[i]
        
        # Get the window of bars before and after
        before_window = df['h'].iloc[i-window:i]
        after_window = df['h'].iloc[i+1:i+window+1]
        
        # Check if current high is greater than all bars before and after
        if all(current_high > before_window) and all(current_high > after_window):
            highs.append(True)
        else:
            highs.append(False)
    
    # Pad the results with False for the bars at the start and end
    highs = [False] * window + highs + [False] * window
    
    return highs

# Apply the swing high detection
gold_csv['swing_high'] = detect_swing_highs(gold_csv, window=5)

# Display the swing highs
swing_highs_df = gold_csv[gold_csv['swing_high'] == True]
print("\nNumber of swing highs detected:", len(swing_highs_df))
print("\nFirst few swing highs:")
print(swing_highs_df[['timestamp', 'h', 'o', 'c']].head())


# ... existing code ...
```

```python
import matplotlib.pyplot as plt

# Take first 400 rows of data for plotting
gold_csv_short = gold_csv.iloc[0:400]

# Get swing highs only for the shortened range
swing_highs_short = gold_csv_short[gold_csv_short['swing_high'] == True]

# Create the plot
plt.figure(figsize=(12, 6))

# Plot price line
plt.plot(range(len(gold_csv_short)), gold_csv_short['c'], 
         label='Price', color='blue', alpha=0.6, linewidth=1)

# Plot swing high points
plt.scatter(swing_highs_short.index, swing_highs_short['h'], 
           color='green', marker='^', s=100, label='Swing Highs')

# Customize the plot
plt.title('Gold Price with Swing Highs (First 400 Minutes)')
plt.xlabel('Minutes from Start')
plt.ylabel('Price')
plt.legend()
plt.grid(True, alpha=0.3)

# Adjust layout to prevent label cutoff
plt.tight_layout()

# Show the plot
plt.show()
```

```python
def detect_swing_lows(df, window=5):
    """
    Detect swing lows in price data
    window: number of bars to look before and after (default=5)
    """
    lows = []
    
    # Loop through the data
    for i in range(window, len(df) - window):
        # Get the current low
        current_low = df['l'].iloc[i]
        
        # Get the window of bars before and after
        before_window = df['l'].iloc[i-window:i]
        after_window = df['l'].iloc[i+1:i+window+1]
        
        # Check if current low is less than all bars before and after
        if all(current_low < before_window) and all(current_low < after_window):
            lows.append(True)
        else:
            lows.append(False)
    
    # Pad the results with False for the bars at the start and end
    lows = [False] * window + lows + [False] * window
    
    return lows

# Apply the swing high detection
gold_csv['swing_low'] = detect_swing_lows(gold_csv, window=5)

# Display the swing lows
swing_lows_df = gold_csv[gold_csv['swing_low'] == True]
print("\nNumber of swing lows detected:", len(swing_lows_df))
print("\nFirst few swing lows:")
print(swing_lows_df[['timestamp', 'l', 'o', 'c']].head())


# ... existing code ...
```
