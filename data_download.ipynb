```python
import requests
import pandas as pd
import time
api_key = "olC7b3Iae2DXrAAAvaTBnGapqSzpMjCW"
symbol = "C:XAUUSD"  # Gold Futures on COMEX
interval = "minute"
from_date = "2021-05-01"
to_date = "2025-06-02"

# Initial URL
base_url = f"https://api.polygon.io/v2/aggs/ticker/{symbol}/range/5/{interval}/{from_date}/{to_date}?adjusted=true&sort=asc&limit=50000&apiKey={api_key}"

all_data = []
i = 0
while base_url:
    response = requests.get(base_url)
    if response.status_code != 200:
        print("Request failed:", response.status_code, response.text)
        break

    json_data = response.json()

    # Check if 'results' exist
    if 'results' in json_data:
        all_data.extend(json_data['results'])

    # Get next page URL if available
    base_url = json_data.get('next_url')
    if base_url:
        base_url += f"&apiKey={api_key}"  # Ensure API key is appended
    i=i+1
    if i%5==0:
        time.sleep(60)

# Convert to DataFrame
if all_data:
    df = pd.DataFrame(all_data)
    df['timestamp'] = pd.to_datetime(df['t'], unit='ms')
    print(df.head())
    df.to_csv('gold_data_5_minutes_completed.csv', index=False)
else:
    print("No data returned.")
```
