```python
import numpy as np
import pandas as pd
import backtrader as backtrader
import yfinance as yf
import os
import ta
```

```python
data = pd.read_csv('GC_1min_sample.csv', index_col='timestamp', parse_dates=True)
data.index.name= 'Time'
data.head(5)
```

```python
for i in range(10,31,10):
  data[f'std deviation {i}'] = data['close'].rolling(window=i).std()
for i in range(10,31,10):
  data[f'moving average {i}'] = data['close'].rolling(window=i).mean()
data['body ratio']= np.abs((data['close'] - data['open'])/( data['high']-data['low']))
result = np.where(data['close']>data['open'],(data['high'] - data['close'])/( data['open']-data['low']),(data['high'] - data['open'])/( data['close']-data['low']))
result = np.where(np.isinf(result), 0, result)
data['upper wick to lower wick ratio']= result
data['Returns'] = (data['close']/data['close'].shift(1))
data['Log Returns'] = np.log(data['Returns'])
data['Returns']=(1-data['Returns'])*100
data.dropna(inplace=True)
```
